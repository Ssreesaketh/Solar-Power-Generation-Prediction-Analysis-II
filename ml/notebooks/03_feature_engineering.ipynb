# 03: Physics-Grounded Feature Engineering

### Theoretical Foundations:
Photovoltaic energy conversion is governed by semiconductor physics:
1. **Irradiance**: Primary driver of electron-hole pair generation.
2. **Cell Thermal Derating**: As panel surface temperature rises above standard test conditions ($25^\circ\text{C}$), crystalline silicon efficiency declines by approximately $-0.4\%/^\circ\text{C}$.
3. **Diurnal Geometry**: Solar elevation changes across the day; cyclical sine and cosine encodings capture time of day smoothly across midnight.
4. **Inverter Saturation**: High noon irradiance causes inverter clipping, modeled via quadratic $\text{IRRADIATION}^2$.
